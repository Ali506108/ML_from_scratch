### Handle missing values

```python

num_cols = ['time_management_score' , 'attendance_percentage' , 'previous_gpa' ]

df[num_cols] = df[num_cols].fillna(df[num_cols].median())

df.head()
```

```python
str_cols = ['parent_education','device_availability','sleep_quality','notes_quality']

df[str_cols] = df[str_cols].fillna('Missing')

print(f"Null values is  {df.isnull().sum()}")
```
