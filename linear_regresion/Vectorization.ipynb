### 3.4.1 Indexing
Elements of vectors can be accessed via indexing and slicing. NumPy provides a very complete set of indexing and slicing capabilities. We will explore only the basics needed for the course here. Reference [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html) for more details.
**Indexing** means referring to *an element* of an array by its position within the array.
**Slicing** means getting a *subset* of elements from an array based on their indices.
NumPy starts indexing at zero so the 3rd element of an vector $\mathbf{a}$ is `a[2]`.

```python
a = np.arange(10)

print(f"np arange : {a}")
print(f"a[2].shape : {a[2].shape} a[2] = {a[2]}, accessing an element returns a scaler")
print(f"a[-1] = {a[-1]}")

try:
    c = a[10]
except Exception as e:
    print("The error message you'll get is ")
    print(e)
```

```python
a = np.arange(10)
print(f"a = {a}")

c = a[2:4]
print(f"c[2:4] = {c}")

c = a[1:9:3]
print(f"c[2:7:2] = {c}")

c = a[4:]
print(f"c[4:] = {c}")

c = a[:4]
print(f"c[:4] = {c}")

c = a[:]
print(f"c[:] = {c}")
```

```python
a = np.array([1,2,3,4,5])

print(f"a = {a}")

b = -a
print(f"b = -a = {b}")

b = sum(a)
print(f"b = sum(a) = {b}")

mean_a = np.mean(a)
print(f"mean of a = {mean_a}")

b=a**2
print(f"b = a**2 : {b}")
```

### 3.4.4 Vector Vector element-wise operations
Most of the NumPy arithmetic, logical and comparison operations apply to vectors as well. These operators work on an element-by-element basis. For example
$$ c_i = a_i + b_i $$

```python
a = np.array([1,2,3,4,5])
b = np.array([1,-2,3,-4,5])

print(f"Binary operations work element wise : {a+b}")
```

```python
c = np.array([1,3,2])

try:
    d = a + c
    print(f"c = {d}")
except Exception as e:
    print(f"The error message you'll get is ")
    print(e)
```

```python
a = np.array([1,3,5,9])

dot = a*5
print(f"Dot product : {dot}")
```

Let's implement our own version of the dot product below:

**Using a for loop**, implement a function which returns the dot product of two vectors. The function to return given inputs $a$ and $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Assume both `a` and `b` are the same shape.

```python
def my_dot(a,b):
    """
    Compute the dot product of two vectors
    :param a:(ndarray (n,)) : input vector
    :param b:b (ndarray (n,)):  input vector with same dimension as a
    :return: X(scaler)
    """

    X = 0
    for i in range(len(a)):
        X = X + a[i]*b[i]
    return X
```

```python
a = np.array([1,3,5,9])
b = np.array([1,-2,3,-4,5])

print(f"Dot product : {my_dot(a,b)}")
```

```python
import time
np.random.seed(1)
a = np.random.rand(10000000)
b = np.random.rand(10000000)

time_dot = time.time()
c = np.dot(a , b)
toc = time.time()

print(f"np.dot(a,b) : {c:4f}")
print(f"Vectorized version duration : {1000*(toc-time_dot):.4f}ms")

tic = time.time()
c = my_dot(a,b)
toc = time.time()

print(f"my_dot(a,b) : {c:4f}")
print(f"Loop version duration : {1000*(toc-tic):.4f}ms")
```
